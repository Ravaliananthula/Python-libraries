##Scatterplot

####A scatter plot shows the relationship between two numerical variables using dots on a graph. Each dot represents one data point.

```python
import numpy as np
import matplotlib.pyplot as plt
india = [1,5,6,2,7,4]
usa = [5,4,1,3,2,6]
plt.scatter(x = india, y = usa)
plt.show()
```

```python
india = [1,5,6,2,7,4]
usa = [5,4,1,3,2,6]
uk = [6,3,5,4,1,7]
plt.figure (figsize = (5,3))
plt.title('scatter plot')
plt.xlabel('India')
plt.ylabel('USA')
plt.scatter(x = india, y = usa, color = 'r', marker = '*')
plt.show()
```

##Line Chart

####A line chart displays data points connected by straight lines to show changes or trends over time.

```python
india = [1,5,6,2,7,4]
usa = [5,4,1,3,2,6]
uk = [6,3,5,4,1,7]
plt.figure(figsize = (5,3))
plt.title('Line Chart')
plt.xlabel('India')
plt.ylabel('USA')
plt.plot(india,usa,color='r',marker='*')
plt.show()
```

```python
india = [1,5,6,2,7,4]
usa = [5,4,1,3,2,6]
uk = [6,3,5,4,1,7]
plt.figure(figsize = (5,3))
plt.title('Line chart')
plt.xlabel('India')
plt.ylabel('USA')
plt.plot(india,usa,color='r',marker='*')
plt.plot(india,uk,color='b',marker='*')
plt.show()
```

##Legend

####A legend explains what different colors, symbols, or lines in a chart represent.

```python
india = [1,5,6,2,7,4]
usa = [5,4,1,3,2,6]
uk = [6,3,5,4,1,7]
plt.figure(figsize = (5,3))
plt.title('legend chart')
plt.xlabel('India')
plt.ylabel('USA')
plt.plot(india,usa,marker = '*',label = 'india-usa')
plt.plot(india,uk,marker = '*', label = 'india-uk')
plt.legend()
plt.show()
```

##Subplot

####Subplots are multiple smaller charts displayed within a single figure.

```python
india = [1,5,6,2,7,4]
usa = [5,4,1,3,2,6]
uk = [6,3,5,4,1,7]
plt.figure(figsize = (5,3))
plt.subplot(1,2,1)
plt.title('india - usa')
plt.plot(india,usa,color = 'r',marker = '*')
plt.subplot(1,2,2)
plt.title('india-uk')
plt.plot(india,uk,color = 'g',marker = '*')
plt.show()
```

```python
india = [1,5,6,2,7,4]
usa = [5,4,1,3,2,6]
uk = [6,3,5,4,1,7]
plt.figure(figsize = (4,4))
plt.subplot(2,1,1)
plt.title('india -usa')
plt.plot(india,usa,color = 'r',marker='*')
plt.subplot(2,1,2)
plt.title('india-uk')
plt.plot(india,uk,color='g',marker='*')
plt.show()
```

##Histogram


####A histogram shows the distribution of numerical data using bars grouped into ranges (bins).

```python
age = [1,4,7,9,3,11,14,21,27,29,31,32,35,39]
plt.figure(figsize =(5,3))
plt.hist(x = age,bins=4,color='g')
plt.show()
```

##Bar chart

####A bar chart compares different categories using rectangular bars.

```python
Company_names = ['Google','Meta','MSFT','IBM']
Revenue = [1200,1400,1000,500]
plt.figure(figsize = (5,3))
plt.bar(Company_names,Revenue,color = 'g')
plt.show()
```

```python
Company_names = ['Google','Meta','MSFT','IBM']
Revenue = [1200,1400,1000,500]
plt.figure(figsize=(5,3))
plt.barh(Company_names,Revenue,color = ['r','g','y','black'])
plt.show()
```

##Pie Charts


####A pie chart shows parts of a whole as slices of a circle.

```python
Company_names = ['Google','Meta','MSFT','IBM']
Revenue = [1200,1400,1000,500]
plt.figure(figsize = (5,3))
plt.pie(Revenue,labels = Company_names)
plt.show()
```

```python
Company_names = ['Google','Meta','MSFT','IBM']
Revenue = [1200,1400,1000,500]
plt.figure(figsize = (5,3))
plt.pie(Revenue,labels = Company_names,shadow = True)
plt.show()
```

```python
Company_names = ['Google','Meta','MSFT','IBM']
Revenue = [1200,1400,1000,500]
plt.figure(figsize=(5,3))
plt.pie(Revenue,labels = Company_names,shadow=True,autopct = '%2f')
plt.show()
```

```python
Company_names = ['Google','Meta','MSFT','IBM']
Revenue = [1200,1400,1000,500]
plt.figure(figsize =(5,3))
plt.pie(Revenue,labels = Company_names,shadow =True,autopct = '%2f',explode=[0.3,0,0.2,0])
plt.show()
```
